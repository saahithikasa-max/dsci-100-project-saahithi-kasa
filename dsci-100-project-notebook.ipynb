```r
### EDIT THIS CODE CELL ###

tidy_womens_march_madness <- women_march_madness_data |>
            select(reg_wins, reg_losses, bid) |>  # selects predictor and response variables necessary for building the classification model
            mutate(bid = as_factor(bid)) |> # Modifies the categorical variable into a factor type for classification later on
            mutate(bid = fct_recode(bid, "At-Large" = "at-large", "Automatic" = "auto")) # Renames the categorical variable types for ease of understanding
# As there is a single value to each cell, one observation per row, and one variable per column, this data set can be deemed tidy
# Thus, no further steps are necessary at the moment. 
head(tidy_womens_march_madness) # shows the first 6 rows of the tidied data set
```

```r
### EDIT THIS CODE CELL FOR PLOT 1 ###
# Plot the distribution of Regional Wins in Your Tidied Data Set
options(repr.plot.width = 15, repr.plot.height = 12)
reg_wins_plot <- tidy_womens_march_madness |>
                 ggplot(aes(x = reg_wins)) +
                 geom_histogram() + 
                 labs(x = "Regional Wins for Schools (Women's March Madness)", 
                      y = "Counts for Regional Wins for Schools (Women's March Madness)",
                      title = "Distribution of Regional Wins for Schools Participating in Women's March Madness") +
                      theme(text = element_text(size = 20))
reg_wins_plot
```

```r
### EDIT THIS CODE CELL FOR PLOT 2 ###
# Use ggplot to create a visualization showing the relative proportions of each category/level for one of categorical variable in your tidied data set.
 # plot categorical variables for bids against regional losses 

options(repr.plot.width = 15, repr.plot.height = 12)

reg_losses_proportions <- tidy_womens_march_madness |>
                          group_by(bid) |>
                          summarize(reg_losses = sum(reg_losses = sum(reg_losses, na.rm = TRUE))) |> # summarize function takes total of all regional loss values for each bid type
                          mutate(proportion_reg_losses = (reg_losses/sum(reg_losses))) # divides total reg_losses values by their sums, generating their respective proportions
reg_losses_proportions
                                    
reg_losses_plot <- reg_losses_proportions |>
                   ggplot(aes(x = bid, y = proportion_reg_losses, fill = bid)) + 
                   geom_bar(stat = "identity") + 
                   labs(x = "Women's March Madness Conference Bid Type", 
                        y = "Proportion of Regional Losses (Women's March Madness)", 
                        fill = "Bid Type", 
                        title = "Proportion of Regional Losses vs. Granted Bid Type (Women's March Madness)") +
                   theme(text = element_text(size = 19))
reg_losses_plot
```

The visualization created below graphs the correlation between two numerical variables, the number of regional wins(reg_wins) and the number of regional_losses(reg_losses), where each point is distinguished with a unique color and shape specific to its bid category (automatic or at-large). This plot is relevant to the analysis because it grants me a stronger understanding of where bid types lie in relation to the numerical variables at hand. For instance, I can see that Automatic bids are more densely condensed towards the upper left region of the graph, where there are fewer regional wins overall. 

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###
# This first visualization will plot reg wins vs reg losses and sort each point by bid type

library(RColorBrewer)
# display.brewer.all(colorblindFriendly = TRUE)
options(repr.plot.width = 17, repr.plot.height = 12)
reg_wins_and_losses_plot <- tidy_womens_march_madness |>
                            ggplot(aes(x = reg_wins, y = reg_losses)) +
                            geom_point(aes(colour = bid, shape = bid)) + 
                            labs(x = "Regional Wins for Schools (Women's March Madness)", 
                                 y = "Regional Losses for Schools (Women's March Madness)", 
                                 title = "Correlation Between Regional Wins and Losses within the Women's March Madness Collegiate Conference", 
                                 colour = "Bid Type", # Used shape and color to make the message at hand more redundant
                                 shape = "Bid Type") +
                                 theme(text = element_text(size = 20)) +
                                 scale_color_brewer(palette = "Paired")
reg_wins_and_losses_plot
```

The following visualization depicts the associated proportions of regional wins for each Conference Bid-Type, which is the categorical variable we are utilizing for classification. The generated plot is relevant to the analysis as it allows me an understanding of whether a larger proportion of wins is associated with a particular bid-type. However, from looking at this graph, we can see that is not the case as both proportions are very close in value, indicating that neither bid type is likely to account for a large portion of regional conference wins. 

```r
# The second visualization will analyze the proportion of regional wins against bid type

library(RColorBrewer)
# display.brewer.all(colorblindFriendly = TRUE)
options(repr.plot.width = 17, repr.plot.height = 12)

reg_wins_and_losses_bar <- tidy_womens_march_madness |>
                           group_by(bid) |>
                           summarize(reg_wins = sum(reg_wins, na.rm = TRUE)) |>
                           mutate(proportion_reg_wins = reg_wins/sum(reg_wins))

reg_wins_and_bid_plot <- reg_wins_and_losses_bar |>
                         ggplot(aes(x = bid, y = proportion_reg_wins, fill = bid)) +
                         geom_bar(stat = "identity") +
                         labs(x = "Women's March Madness Conference Bid Type", 
                              y = "Proportion of Regional Wins (Women's March Madness)", 
                              title = "Proportion of Regional Wins vs. Granted Bid Type (Women's March Madness)",
                              fill = "Conference Bid Type") +
                         theme(text = element_text(size = 19))
reg_wins_and_bid_plot
```
